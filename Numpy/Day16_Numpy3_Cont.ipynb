```python
# Vector Operations : collection of scalars
first_vector = np.array([1,2,3,4,5])
second_vector = np.array([6,7,8,9,10])
print(first_vector + second_vector)  # vector + vector = vector
print(first_vector - second_vector)
print(first_vector * second_vector)
print(first_vector / second_vector)
print(first_vector ** second_vector)
```

```python
# array multiplication
first_vector = np.array([1,2,3])
second_vector = np.array([4,5,6])
# result = first_vector.dot(second_vector)
# print("Dot Product:", result)   
result_vector = np.multiply(first_vector, second_vector)
print("Element-wise Multiplication:", result_vector)    
```

```python
# matrix operations

first_matrix = np.array([[1,2,3],[4,5,6]])
second_matrix = np.array([[7,8,9],[10,11,12]])
print("Matrix Addition:\n", first_matrix + second_matrix)   
print("Matrix Subtraction:\n", first_matrix - second_matrix)
print("Matrix Element-wise Multiplication:\n", first_matrix * second_matrix)
print("Matrix Element-wise Division:\n", first_matrix / second_matrix)
```

```python
# matrix multiplication 

# number of column in 1st matrix = number of rows in 2nd matrix
first_matrix = np.array([1,2,3])
second_matrix = np.array ([12,13,14])
print(first_matrix)
print(second_matrix)

# Array multiplication: 
array_mul = np.multiply(first_matrix , second_matrix)
print("array multiplication is :\n" , array_mul)

# matrix multiplication :
matrix_mul = np.matmul(first_matrix , second_matrix)
print("matrix multiplication is :\n" , matrix_mul)
```

```python
new_array = np.array([[1,2,3] , [4,5,6]])
print ("New Array :\n" , new_array)
print(new_array.shape)
now_array = np.array([[10,20,30], [40,50,60]])  
print ("Now Array:\n", now_array)
print(now_array.shape)
```

```python
# performing matrix multiplication

matrix_mul = np.matmul(new_array , now_array)
print("matrix multiplication:\n" , matrix_mul)
```

```python
# transpose of matrix

tranposed_array = np.transpose(now_array)
print("Transposed array :\n" , tranposed_array)
print(tranposed_array.shape)
```

```python
# performing  matrix multiplication 

matrix_mul = np.matmul(new_array , tranposed_array)
print("matrix multiplication:\n" , matrix_mul)
```
