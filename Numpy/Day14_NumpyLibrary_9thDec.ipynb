# Que :  what is the Data Structure of Numpy
Ans : the Data Structure of Numpy as know as nd array - which means : n -dimensional array 

```python
# Creating an array in Numpy
arr=np.array([1,2,3,4,5]) # group of scalars
print(arr)
print(type(arr))
```

# A single number is know as : scalar EX: [5]   ----> has no Dimension [0-Zero]

# A group of scalars is known as vector EX: [1,2,3,4,5] ----> has min 1 Dimension [1-D]

# A group of vectors is know as a Matrix  EX: [ [1,2,3,4,5] , [6,7,8,9,] ] ----> has min of 2 Dimension [2-D]

# A group of Matrix is know as Tensor EX: [] ----> has min of 3 dimension [3-D]

```python
# Finding more information about a numpy arfray [a- attributes & methods of numpy array]
print("data type of array arr :", arr.dtype)
print("Number of dimension of array arr:" , arr.ndim)
print("Shape of array arr:", arr.shape) # number of rows and columns in the array
print("Size of array arr:", arr.size)   # total number of elements in the array
print("item size of array arr:", arr.itemsize) # size of each element in the array in bytes
print("nbytes of array arr:", arr.nbytes) # total size of the array in bytes (size * itemsize)


print("strides of array arr:", arr.strides) # number of bytes to step in each dimension when traversing an array
print("flags of array arr:", arr.flags) # memory layout of the array
print("base of array arr:", arr.base) # base object if the array is a view of another array
print("T (transpose) of array arr:", arr.T) # transpose of the array
print("flat of array arr:", arr.flat) # flat iterator over the array
print("ctypes of array arr:", arr.ctypes) # ctypes interface of the array
print("data of array arr:", arr.data) # buffer containing the actual elements of the array
print("tolist of array arr:", arr.tolist()) # convert array to a (possibly nested) list
print("view of array arr:", arr.view()) # new view of the array with the same data
print("copy of array arr:", arr.copy()) # new array object with a copy of the data
```
