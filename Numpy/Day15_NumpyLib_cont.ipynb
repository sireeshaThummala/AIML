```python
# Reshape command in numpy
arr = np.arange(1,13) # n-1
print("Original Array:\n", arr)
reshaped_arr = arr.reshape(3,4) # 3 rows and 4 columns
print("Reshaped array (3*4):\n",reshaped_arr)
print(arr.ndim)
print(arr.shape)
```

```python
arr.reshape(4,3)
print("reshaped array is:\n", arr.reshape(4,3))
print(arr.ndim) # number of dimensions
print(arr.shape)
```

```python
arr.reshape(2,2,3)
print("reshaped array is:\n", arr.reshape(2,2,3))
print(arr.ndim) # number of dimensions
print(arr.shape)
```

```python
# Linspace : To create an array with evenly spaced values
# Syntax: np.linspace(start, stop, num , endpoint , retstep , dtype)

linspace_array = np.linspace(0, 10, 5, endpoint=False , retstep=True) # 5 values between 0 and 10
```

```python
lin_arr_1 =np.linspace(0, 1, 5)  # 0 to 1 , 5 values
print("Linspace array:\n", linspace_array)
print("Linspace array 1:\n", lin_arr_1)
```

```python
lin_arr_2 =np.linspace(1,100, 50) 
print("Linspace array 2:\n", lin_arr_2)
```
