# In python we have 3 types of conditional statments

1. if statment
   # if statment can also be considered for decision making only.
   # If statment works only if the given condition is true. 
   # Decision making purpose


2. for loop
   # For Loop will not validate if the hgiven condition is true or not 
   # just to repeat / itereat anything for n number of times


3. while loop
   # while Loop runs till the given condition is true. 
   # while Loop runs till 

```python
# if-elif-else example
A = 20
B = 20

if A > B:
    print("A is greater than B")
elif A < B :
    print("A is less than B")
else:
    print("A is equal B")       
```

```python
# Working with For Loop
# Writing 2 table using for loop
# sync : for()

# range function : generate a sequence of numbers { start , stop }

for i in range(1,11):
    print(f"2 * {i} = {2 * i}")  # formated string literals ( f-strings)
```

```python
# Working with range function

# Range function purpose: to generate a sequence of numbers (start, stop)
# printing 1 to 100 numbers using range function
for c in range(1,5):  # c just store the sequence of numbers (n-1)
    print(c)
```

```python
# Working with While loops : while loop runs till the given condition is met

a = 1
b = 5
while a<=b:
    print(a)
    a += 1
    
```

```python
# create a python program to calculate all of your expences and print the total amount spent in a month

# food travel purchases: for 30 0r 31 days 
total = 0

while True:
    expense = input("Enter an expense amount (or type 'done'): ")

    if expense.lower() == "done":
        break

    if expense.isdigit():      # check valid number
        total += int(expense)
    else:
        print("Invalid input!")

print("Total Monthly Expenses:", total)
```
