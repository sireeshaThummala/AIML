```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error,root_mean_squared_error , classification_report, confusion_matrix
```

```python
# Mean Absolute Error (MAE)
# Definition : It bis the Avg of the absolute difference between predicted and actual values
# Purpose / where do use : Linear Regrtession , Decision Tree, Random Foret Regression
# Import Statment : sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error,root_mean_squared_error
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error,root_mean_squared_error
```

```python
y_true = np.array([3, -0.5 , 2, 7])
y_pred = np.array([2.5 , 0.0 , 2, 8])
mae = mean_absolute_error(y_true , y_pred)
print ("Mean Absolute Error :",mae)
```

```python
# Mean Squared Error (MSE)
# From sklearn.metrics import mean_squared_error
y_true = np.array([3, -0.5 , 2, 7])
y_pred = np.array([2.5 , 0.0 , 2, 8])
mse = mean_squared_error(y_true , y_pred)
print ("Mean Absolute Error :",mse)
```

```python
# Root Mean Squared Error (RMSE)
y_true = np.array([3, -0.5 , 2, 7])
y_pred = np.array([2.5 , 0.0 , 2, 8])
rmse = np.sqrt(mean_squared_error(y_true , y_pred))
print ("Mean Absolute Error :",rmse)
```

```python
# MAE , MSE , RMSE , r2_score
# r2_score :
# purpose: It meaures how well the model explains the variance in the target vriable . How well the relationship between 
# dependent and independent variables.   
# when we make prediction at the time instead of prediction , we will use the word score( real)
y_true = np.array([3, -0.5 , 2, 7])
y_pred = np.array([2.5 , 0.0 , 2, 8])
r2 = r2_score(y_true , y_pred)
print ("R Squared:",r2)
```

```python
# MAE , MSE , RMSE , r2_score
# Classification Report : It is used to eveluate the performancxe of a classifiction model.
#inmport statment : from sklearn.metrics import classification_report
from sklearn.metrics import classification_report
y_true = np.array([1,0,1,1])
y_pred = np.array([0,0,1,1])
report = classification_report(y_true , y_pred)
print("Classification Report :\n" , report)
```
