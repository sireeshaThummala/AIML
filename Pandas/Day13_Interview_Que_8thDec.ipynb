```python
# explain deep and shallow copy in python?
# Shallow Copy: Creates a new object but inserts references into it to the objects found in the original. copies memory address
# Changes to mutable objects in the copy will affect the original.
# Deep Copy: Creates a new object and recursively adds copies of nested objects found in the original
# Changes to mutable objects in the copy will not affect the original.
# How to handle categorical data in pandas?
# Methods: astype('category'), get_dummies(), CategoricalDtype
# Use Cases: Memory optimization, Encoding categorical variables for machine learning
```
