## 1. Series

A **Series** is a one-dimensional array with labeled axes (index). It's similar to a list but with named index labels.

### Creating a Series:

```python
import pandas as pd
```

```python
# Method 1: Creating a Series from a list
data_list = [10, 20, 30, 40, 50]
series1 = pd.Series(data_list)
print("Series from list:")
print(series1)
print("\n")
```

```python
# Method 2: Creating a Series with custom index
data_dict = {'A': 100, 'B': 200, 'C': 300, 'D': 400}
series2 = pd.Series(data_dict)
print("Series from dictionary:")
print(series2)
print("\n")
```

```python
# Method 3: Creating a Series with custom index labels
data_list = [10, 20, 30, 40, 50]
series3 = pd.Series(data_list, index=['a', 'b', 'c', 'd', 'e'])
print("Series with custom index:")
print(series3)
print("\n")

# Accessing elements from Series
print("Accessing by index label:")
print(f"Value at index 'c': {series3['c']}")
print("\nAccessing by position (iloc):")
print(f"Value at position 0: {series3.iloc[0]}")
print("\n")
```

### Series Operations and Attributes

```python
# Series attributes
print("Series Information:")
print(f"Series shape: {series3.shape}")
print(f"Series dtype: {series3.dtype}")
print(f"Series index: {series3.index.tolist()}")
print(f"Series values: {series3.values}")
print("\n")

# Series operations
print("Series Operations:")
print(f"Sum: {series3.sum()}")
print(f"Mean: {series3.mean()}")
print(f"Min: {series3.min()}")
print(f"Max: {series3.max()}")
print(f"Standard Deviation: {series3.std()}")
print("\n")
```

## 2. DataFrame

A **DataFrame** is a two-dimensional data structure with labeled axes (rows and columns). It's similar to a spreadsheet or SQL table.

### Creating a DataFrame:

```python
# Method 1: Creating a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28],
    'Salary': [50000, 60000, 75000, 55000],
    'City': ['New York', 'London', 'Paris', 'Berlin']
}
df1 = pd.DataFrame(data)
print("DataFrame from dictionary:")
print(df1)
print("\n")
```

```python
# Method 2: Creating a DataFrame from a list of lists
data_list = [['Alice', 25, 50000], ['Bob', 30, 60000], ['Charlie', 35, 75000]]
df2 = pd.DataFrame(data_list, columns=['Name', 'Age', 'Salary'])
print("DataFrame from list of lists:")
print(df2)
print("\n")
```

```python
# Method 3: Creating a DataFrame from a CSV file (using titanic.csv)
df3 = pd.read_csv('titanic.csv')
print("DataFrame from CSV file (first 5 rows):")
print(df3.head())
print("\n")
```

### DataFrame Filtering and Data Manipulation

```python
# DataFrame information and properties
print("DataFrame Shape:", df1.shape)  # Returns (rows, columns)
print("Column Names:", df1.columns.tolist())
print("Data Types:")
print(df1.dtypes)
print("\nBasic DataFrame Info:")
print(df1.info())
print("\n")
```

```python
# Accessing data in DataFrame
print("Accessing columns:")
print(df1['Name'])  # Access a single column
print("\nAccessing rows using iloc:")
print(df1.iloc[0])  # Access first row
print("\nAccessing rows using loc:")
print(df1.loc[0])   # Access row by label
print("\n")
```

## 3. Data Handling & Cleaning

### Dealing with Missing Data

```python
# Filtering: Select rows where Age > 28
filtered_df = df1[df1['Age'] > 28]
print("Employees with Age > 28:")
print(filtered_df)
print("\n")

# Select multiple columns
selected_cols = df1[['Name', 'Age']]
print("Selected columns (Name and Age):")
print(selected_cols)
print("\n")

# Add a new column
df1['Bonus'] = df1['Salary'] * 0.10
print("DataFrame with new Bonus column:")
print(df1)
print("\n")
```

### DataFrame Statistical Operations

```python
# Statistical operations on DataFrame
print("Statistical Summary:")
print(df1.describe())
print("\n")

# Mean of numeric columns
print("Mean values:")
print(df1.mean(numeric_only=True))
print("\n")

# Sum of numeric columns
print("Sum values:")
print(df1.sum(numeric_only=True))
print("\n")
```

```python
# Creating a DataFrame with missing values
import numpy as np
df_missing = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, np.nan, 35, 28, np.nan],
    'Salary': [50000, 60000, np.nan, 55000, 75000]
})
print("DataFrame with Missing Values:")
print(df_missing)
print("\n")

# Check for missing values
print("Missing Values Count:")
print(df_missing.isnull().sum())
print("\n")

# Drop rows with missing values
df_dropped = df_missing.dropna()
print("After dropping rows with missing values:")
print(df_dropped)
print("\n")

# Fill missing values with specific values
df_filled = df_missing.fillna({'Age': df_missing['Age'].mean(), 'Salary': 65000})
print("After filling missing values:")
print(df_filled)
print("\n")
```

### Sorting and Grouping Data

```python
# Sorting DataFrame by a column
df_sorted = df1.sort_values(by='Age')
print("DataFrame sorted by Age (ascending):")
print(df_sorted)
print("\n")

# Sorting in descending order
df_sorted_desc = df1.sort_values(by='Salary', ascending=False)
print("DataFrame sorted by Salary (descending):")
print(df_sorted_desc)
print("\n")

# Grouping data
print("Grouping by Age and calculating mean salary:")
grouped = df1.groupby('Age')['Salary'].mean()
print(grouped)
print("\n")
```

## 4. Advanced DataFrame Operations

### Concatenation and Merging

```python
# Concatenating DataFrames
df_part1 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Age': [25, 30]})
df_part2 = pd.DataFrame({'Name': ['Charlie'], 'Age': [35]})
df_concatenated = pd.concat([df_part1, df_part2], ignore_index=True)
print("Concatenated DataFrame:")
print(df_concatenated)
print("\n")

# Merging DataFrames on a common column
df_employees = pd.DataFrame({'EmployeeID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
df_departments = pd.DataFrame({'EmployeeID': [1, 2, 3], 'Department': ['HR', 'IT', 'Finance']})
df_merged = pd.merge(df_employees, df_departments, on='EmployeeID')
print("Merged DataFrame:")
print(df_merged)
print("\n")
```
