# In General we use some of the following charts / plots / graphs

# Type of charts :

# 1. Line Plot : To show trend in data
# 2. Bar Plot : To comparing difference between diff categories
# 3. Pie pLot : To comparing difference between diff categories
# 4. Histogram : To show distribution of data
# 5. Scatter Plot : To show relationship between 2 variables
# 6. Box plot : To show distribution of data and outliers
# 7. Violin Plot : To show distribution of data and outliers
# 8. Heatmap : To show corrwlation between variables : works only on numerical data

```python
#  Working with Line Plot : 1) purpose of Plot.  (2) Required variables/ parameters for each type of plot

# purpose of plot : to show trend in data
# required variables / parameters:  x-axis , y-axis , data
# x-axis : Age
#y-axis: Fare
# data: df
# to show trend in data : line plot
# x-axis : Age
#y-axis: Fare
# data: df

sns.lineplot(x='age' , y='fare' , data=df)

```

```python
# creating a line plot with matplotlib with same column x= age , y = fare
plt.plot(df['age'] , df["fare"])
plt.show()
```

```python
# Working with Bar Plot : horizontal bar , vertical var
sns.barplot(x='age' , y='fare', data=df)
```

```python
# Working with Barplot in matplotlib 
plt.bar(df["age"], df["fare"])
plt.show
```

```python
#creating with bar plot horzontal
plt.barh(df["age"], df["fare"])
plt.show()
```

```python
# working with Histogram Plot : to show distribution of data

plt.hist(df["fare"])
plt.show()

```

```python
# trying histogram Plot : to show distribution of data
# it takes only one parameter
plt.hist(df["fare"])
plt.show()
```

```python
# Trying histogram with bins

plt.hist(df['fare'] , bins=10)
plt.show()
```

```python
# Using seaborn to create histogram plot

sns.histplot(df['fare'])
plt.show()
```

```python
# Working with scatter plot

sns.scatterplot(df["age"])
plt.show()
```

```python
# trying scatter plot with matplotlib
plt.scatter(df['age'],df['fare'])
plt.show()
```

```python
# working with violin plot
plt.violinplot(df['age'] , df['fare'])
plt.show()
```

```python
sns.violinplot(df['fare'])
plt.show()
```

```python
#working with box plot
sns.boxplot(df['fare'])
```

```python
plt.boxplot(df['fare'])
plt.show()
```

```python
# working with heatmap
sns.heatmap(df.corr())
plt.show()
```
